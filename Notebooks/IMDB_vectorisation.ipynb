```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('/content/imdb_text_preprocessed.csv')
print(df.info())
print(df.head())
```

```python
import ast
text_str = df['review'][2]
text_list = np.array(ast.literal_eval(text_str))
print(text_list.reshape(-1, 1))
```

```python
#ohe impolementaion
from sklearn.preprocessing import OneHotEncoder
onehotencoder = OneHotEncoder()

sample_text_ohe = onehotencoder.fit_transform(text_list.reshape(-1, 1))
print(sample_text_ohe.toarray())
```

```python
#n-grams implementation(uni and bi)

from sklearn.feature_extraction.text import CountVectorizer
cv_uni = CountVectorizer()
sample_text_uni = cv_uni.fit_transform(text_list)
print(sample_text_uni.toarray())
```

```python
print(cv_uni.vocabulary_)
```

```python
#bigram
#sample text for bi gram
def convert_to_string(text_list):
  sample_text_str = ""
  for words in text_list:
    sample_text_str = sample_text_str + " "+ words
  return sample_text_str

for i in range(0,3):
  doc_li = []
  text_str_bi = df['review'][i]
  text_list_bi = np.array(ast.literal_eval(text_str))
  sample_text_str = convert_to_string(text_list)
  doc_li.append(sample_text_str)
```

```python
cv_bi = CountVectorizer(ngram_range=(2, 2))
sample_text_bi = cv_bi.fit_transform(doc_li)
print(sample_text_bi.toarray())
```

```python
print(cv_bi.vocabulary_)
```

```python
#tf-idf implementation
from sklearn.feature_extraction.text import TfidfVectorizer
tfidf = TfidfVectorizer()

sample_text_tfidf = tfidf.fit_transform(doc_li)
print(sample_text_tfidf.toarray())
```

```python
#vectorising the whole df using tf-idf

all_docs=[]
for review in df['review']:
  text_list = np.array(ast.literal_eval(review))
  sample_text_str = convert_to_string(text_list)
  all_docs.append(sample_text_str)

tfidf_all = TfidfVectorizer()
final_df_all = tfidf_all.fit_transform (all_docs)

print(final_df_all.shape)
```

```python
#generating a sample for 1000 rows to convert it to tfidf vector

df_10000 = df.head(10000).copy()
df_10000['tfidf_vector'] = list(final_df_all[:10000].toarray())
```
